```python

# ================================================================
# HOUSING PRICE PREDICTION PROJECT
# Multiple Linear Regression + RFE + Backward Elimination
# Dataset: Housing.csv
# Platform: Google Colab
# ================================================================

# ================================================================
# 1. IMPORT NECESSARY LIBRARIES
# ================================================================

import os
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import statsmodels.api as sm

from google.colab import files

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler
from sklearn.linear_model import LinearRegression
from sklearn.feature_selection import RFE
from sklearn.metrics import (
    r2_score,
    mean_absolute_error,
    mean_squared_error
)

from statsmodels.stats.outliers_influence import variance_inflation_factor

warnings.filterwarnings("ignore")
sns.set_theme(style="whitegrid")

print("Libraries imported successfully.")


# ================================================================
# 2. IMPORT THE DATASET
# ================================================================

# Upload Housing.csv if it is not already present in Colab.

if not os.path.exists("/content/Housing.csv"):
    print("Please upload your Housing.csv file.")
    uploaded = files.upload()

    if not uploaded:
        raise ValueError("No dataset uploaded.")

    csv_files = [
        name for name in uploaded
        if name.lower().endswith(".csv")
    ]

    if not csv_files:
        raise ValueError("Please upload a CSV file.")

    dataset_path = csv_files[0]
else:
    dataset_path = "/content/Housing.csv"

housing = pd.read_csv(dataset_path)

# Clean accidental spaces in column names.
housing.columns = housing.columns.str.strip()

print("\nDataset loaded successfully.")
print("Dataset shape:", housing.shape)

print("\nFirst five rows:")
display(housing.head())

print("\nDataset information:")
housing.info()

print("\nNumerical summary:")
display(housing.describe())

print("\nCategorical summary:")
display(housing.describe(include="object"))

print("\nMissing values:")
print(housing.isnull().sum())

print("\nDuplicate rows:", housing.duplicated().sum())

# Remove duplicate records, if any.
housing = housing.drop_duplicates().reset_index(drop=True)

# Verify required columns.
required_columns = [
    "price", "area", "bedrooms", "bathrooms", "stories",
    "mainroad", "guestroom", "basement", "hotwaterheating",
    "airconditioning", "parking", "prefarea", "furnishingstatus"
]

missing_columns = [
    col for col in required_columns
    if col not in housing.columns
]

if missing_columns:
    raise ValueError(
        f"Required columns are missing: {missing_columns}"
    )

# Remove records with missing values for this project.
housing = housing.dropna().reset_index(drop=True)

# Ensure target is numeric.
housing["price"] = pd.to_numeric(
    housing["price"], errors="raise"
)

print("\nFinal dataset shape:", housing.shape)


# ================================================================
# 3. EXPLORATORY DATA ANALYSIS (EDA)
# ================================================================

numeric_columns = [
    "price", "area", "bedrooms",
    "bathrooms", "stories", "parking"
]

print("\nNumerical feature distributions:")

housing[numeric_columns].hist(
    figsize=(14, 9),
    bins=20,
    edgecolor="black"
)

plt.suptitle("Distribution of Numerical Variables", fontsize=16)
plt.tight_layout()
plt.show()


# Pairplot of numerical variables
sns.pairplot(
    housing[numeric_columns],
    diag_kind="hist"
)

plt.suptitle("Pairplot of Housing Dataset", y=1.02)
plt.show()


# Correlation heatmap
plt.figure(figsize=(10, 7))

sns.heatmap(
    housing[numeric_columns].corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap of Numerical Variables")
plt.tight_layout()
plt.show()


# Categorical features versus price
categorical_columns = [
    "mainroad", "guestroom", "basement",
    "hotwaterheating", "airconditioning",
    "prefarea", "furnishingstatus"
]

fig, axes = plt.subplots(3, 3, figsize=(17, 13))
axes = axes.flatten()

for i, col in enumerate(categorical_columns):
    sns.boxplot(
        data=housing,
        x=col,
        y="price",
        ax=axes[i]
    )
    axes[i].set_title(f"{col} vs Price")
    axes[i].tick_params(axis="x", rotation=15)

for j in range(len(categorical_columns), len(axes)):
    axes[j].axis("off")

plt.tight_layout()
plt.show()


# Price versus area
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=housing,
    x="area",
    y="price"
)

plt.title("House Area vs Price")
plt.tight_layout()
plt.show()


# Price versus number of bathrooms
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=housing,
    x="bathrooms",
    y="price"
)

plt.title("Number of Bathrooms vs Price")
plt.tight_layout()
plt.show()


# ================================================================
# 4. DATA PREPROCESSING AND FEATURE ENGINEERING
# ================================================================

# Separate the target and independent variables.
X_raw = housing.drop(columns=["price"]).copy()
y = housing["price"].copy()

# Standardize text values in categorical columns.
for col in categorical_columns:
    X_raw[col] = (
        X_raw[col]
        .astype(str)
        .str.strip()
        .str.lower()
    )

# Convert yes/no variables to 1/0.
binary_columns = [
    "mainroad", "guestroom", "basement",
    "hotwaterheating", "airconditioning", "prefarea"
]

for col in binary_columns:
    invalid_values = set(X_raw[col].unique()) - {"yes", "no"}

    if invalid_values:
        raise ValueError(
            f"Unexpected values in {col}: {invalid_values}"
        )

    X_raw[col] = X_raw[col].map({"yes": 1, "no": 0})


# One-hot encode furnishing status.
# Drop the first category to avoid redundant dummy variables.
X = pd.get_dummies(
    X_raw,
    columns=["furnishingstatus"],
    drop_first=True,
    dtype=int
)

# Ensure furnishing status has the expected reference category.
expected_dummy_columns = [
    "furnishingstatus_semi-furnished",
    "furnishingstatus_unfurnished"
]

for col in expected_dummy_columns:
    if col not in X.columns:
        X[col] = 0

# Fix a stable column order.
feature_columns = [
    "area", "bedrooms", "bathrooms", "stories",
    "mainroad", "guestroom", "basement",
    "hotwaterheating", "airconditioning", "parking",
    "prefarea",
    "furnishingstatus_semi-furnished",
    "furnishingstatus_unfurnished"
]

X = X[feature_columns].astype(float)

# Split before fitting the scaler.
X_train_raw, X_test_raw, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.30,
    random_state=100
)

# Scale only numerical predictors.
# The target price is NOT scaled.
scale_columns = [
    "area", "bedrooms", "bathrooms", "stories", "parking"
]

scaler = MinMaxScaler()

X_train = X_train_raw.copy()
X_test = X_test_raw.copy()

X_train.loc[:, scale_columns] = scaler.fit_transform(
    X_train_raw[scale_columns]
)

X_test.loc[:, scale_columns] = scaler.transform(
    X_test_raw[scale_columns]
)

# Keep the exact feature order consistent.
X_train = X_train[feature_columns]
X_test = X_test[feature_columns]

print("\nTraining feature shape:", X_train.shape)
print("Testing feature shape:", X_test.shape)
print("Training target shape:", y_train.shape)
print("Testing target shape:", y_test.shape)

print("\nEncoded and scaled training features:")
display(X_train.head())

print("\nTraining target:")
display(y_train.head())


# ================================================================
# 5. BUILD THE LINEAR MODEL USING RFE AND STATSMODELS
# ================================================================

# ------------------------------------------------
# 5A. Recursive Feature Elimination (RFE)
# ------------------------------------------------

linear_model = LinearRegression()

number_of_features = min(10, X_train.shape[1])

rfe = RFE(
    estimator=linear_model,
    n_features_to_select=number_of_features,
    step=1
)

rfe.fit(X_train, y_train)

selected_features = X_train.columns[rfe.support_].tolist()

print("\n========== RFE RESULTS ==========")

rfe_results = pd.DataFrame({
    "Feature": X_train.columns,
    "Selected": rfe.support_,
    "Ranking": rfe.ranking_
}).sort_values(
    by=["Selected", "Ranking"],
    ascending=[False, True]
)

display(rfe_results)

print("\nFeatures selected by RFE:")
print(selected_features)

print("\nFeatures eliminated by RFE:")
print(rfe_results.loc[~rfe_results["Selected"], "Feature"].tolist())


# ------------------------------------------------
# 5B. OLS MODEL AND BACKWARD ELIMINATION
# ------------------------------------------------

# Start with the features selected by RFE.
active_features = selected_features.copy()

SIGNIFICANCE_LEVEL = 0.05
VIF_THRESHOLD = 5.0

def fit_ols(features, X_data, y_data):
    """Fit an OLS regression model with an intercept."""
    design_matrix = sm.add_constant(
        X_data[features],
        has_constant="add"
    )

    return sm.OLS(y_data, design_matrix).fit()


def calculate_vif(X_data, features):
    """Calculate VIF for predictors, excluding the intercept."""
    if len(features) < 2:
        return pd.DataFrame({
            "Feature": features,
            "VIF": [1.0] * len(features)
        })

    design_matrix = sm.add_constant(
        X_data[features],
        has_constant="add"
    )

    results = []

    for i, feature in enumerate(features, start=1):
        try:
            vif_value = variance_inflation_factor(
                design_matrix.values,
                i
            )
        except Exception:
            vif_value = np.inf

        results.append({
            "Feature": feature,
            "VIF": vif_value
        })

    return pd.DataFrame(results).sort_values(
        by="VIF",
        ascending=False
    )


# Backward elimination:
# 1. Remove the feature with the largest p-value above 0.05.
# 2. Once all p-values meet the threshold, inspect VIF.
# 3. Remove the highest-VIF predictor if VIF exceeds 5.
# 4. Refit the model after every removal.

removal_log = []

while True:

    ols_model = fit_ols(
        active_features,
        X_train,
        y_train
    )

    # Check statistical significance.
    feature_pvalues = ols_model.pvalues.drop(
        labels=["const"],
        errors="ignore"
    )

    insignificant = feature_pvalues[
        feature_pvalues > SIGNIFICANCE_LEVEL
    ]

    if not insignificant.empty:
        feature_to_remove = insignificant.idxmax()

        removal_log.append({
            "Feature Removed": feature_to_remove,
            "Reason": (
                f"P-value = {feature_pvalues[feature_to_remove]:.6f}"
            )
        })

        active_features.remove(feature_to_remove)

        if not active_features:
            raise RuntimeError(
                "All predictors were removed. "
                "Check the dataset and selection criteria."
            )

        continue

    # Check multicollinearity after significance pruning.
    vif_table = calculate_vif(X_train, active_features)

    excessive_vif = vif_table[
        vif_table["VIF"] > VIF_THRESHOLD
    ]

    if not excessive_vif.empty and len(active_features) > 1:
        feature_to_remove = excessive_vif.iloc[0]["Feature"]
        vif_value = excessive_vif.iloc[0]["VIF"]

        removal_log.append({
            "Feature Removed": feature_to_remove,
            "Reason": f"VIF = {vif_value:.4f}"
        })

        active_features.remove(feature_to_remove)
        continue

    # Both checks have passed.
    break


# Final OLS model
final_model = fit_ols(
    active_features,
    X_train,
    y_train
)

print("\n========== FEATURES REMOVED ==========")

if removal_log:
    display(pd.DataFrame(removal_log))
else:
    print("No features were removed.")

print("\n========== FINAL OLS MODEL SUMMARY ==========")
print(final_model.summary())

print("\nFinal selected features:")
print(active_features)

print("\nFinal training R-squared:", round(final_model.rsquared, 4))
print(
    "Final adjusted R-squared:",
    round(final_model.rsquared_adj, 4)
)

print("\nFinal p-values:")
print(final_model.pvalues)

print("\nFinal VIF table:")
final_vif = calculate_vif(X_train, active_features)
display(final_vif)


# ================================================================
# 6. RESIDUAL ANALYSIS OF TRAINING DATA
# ================================================================

X_train_final = sm.add_constant(
    X_train[active_features],
    has_constant="add"
)

y_train_pred = final_model.predict(X_train_final)

residuals = y_train - y_train_pred

# Residual distribution
plt.figure(figsize=(9, 5))

sns.histplot(
    residuals,
    kde=True,
    bins=25
)

plt.axvline(0, linestyle="--")
plt.title("Distribution of Training Residuals")
plt.xlabel("Residual (Actual Price - Predicted Price)")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()


# Residuals against fitted values
plt.figure(figsize=(9, 5))

plt.scatter(
    y_train_pred,
    residuals,
    alpha=0.7
)

plt.axhline(0, linestyle="--")
plt.title("Residuals vs Fitted Values")
plt.xlabel("Fitted Price")
plt.ylabel("Residual")
plt.tight_layout()
plt.show()


# Q-Q plot for checking residual normality
sm.qqplot(
    residuals,
    line="45",
    fit=True
)

plt.title("Q-Q Plot of Training Residuals")
plt.tight_layout()
plt.show()

print("\nMean training residual:", round(residuals.mean(), 4))


# ================================================================
# 7. PREDICTION ON THE TEST DATA
# ================================================================

X_test_final = sm.add_constant(
    X_test[active_features],
    has_constant="add"
)

# Ensure the test design matrix matches the training model.
X_test_final = X_test_final.reindex(
    columns=final_model.model.exog_names,
    fill_value=0
)

y_pred = final_model.predict(X_test_final)

prediction_results = pd.DataFrame({
    "Actual Price": y_test,
    "Predicted Price": y_pred,
    "Prediction Error": y_test - y_pred
})

print("\n========== TEST PREDICTIONS ==========")
display(prediction_results.head(15))


# ================================================================
# 8. MODEL EVALUATION
# ================================================================

r2 = r2_score(y_test, y_pred)
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)

print("\n========== MODEL EVALUATION ==========")
print(f"Test R-squared (R2): {r2:.4f}")
print(f"Mean Absolute Error (MAE): {mae:,.2f}")
print(f"Mean Squared Error (MSE): {mse:,.2f}")
print(f"Root Mean Squared Error (RMSE): {rmse:,.2f}")

print(
    "\nTest R-squared as percentage:",
    f"{r2 * 100:.2f}%"
)


# Actual vs predicted scatter plot
plt.figure(figsize=(8, 6))

plt.scatter(
    y_test,
    y_pred,
    alpha=0.7
)

min_price = min(y_test.min(), y_pred.min())
max_price = max(y_test.max(), y_pred.max())

plt.plot(
    [min_price, max_price],
    [min_price, max_price],
    linestyle="--"
)

plt.title("Actual vs Predicted House Prices")
plt.xlabel("Actual Price")
plt.ylabel("Predicted Price")
plt.tight_layout()
plt.show()


# Compare actual and predicted prices for the first 30 test records
comparison = prediction_results.head(30)

plt.figure(figsize=(13, 5))

plt.plot(
    range(len(comparison)),
    comparison["Actual Price"].values,
    marker="o",
    label="Actual Price"
)

plt.plot(
    range(len(comparison)),
    comparison["Predicted Price"].values,
    marker="x",
    label="Predicted Price"
)

plt.title("Actual vs Predicted Prices: First 30 Test Records")
plt.xlabel("Test Record")
plt.ylabel("House Price")
plt.legend()
plt.tight_layout()
plt.show()


# ================================================================
# 9. FINAL MODEL AS A MATHEMATICAL EQUATION
# ================================================================

print("\n========== FINAL REGRESSION EQUATION ==========")

intercept = final_model.params["const"]

print(f"Predicted Price = {intercept:,.4f}")

for feature in active_features:
    coefficient = final_model.params[feature]

    if coefficient >= 0:
        print(f"    + ({coefficient:,.4f} * {feature})")
    else:
        print(f"    - ({abs(coefficient):,.4f} * {feature})")

print(
    "\nIMPORTANT: The numerical predictors in this equation "
    "are MinMax-scaled values, not their original raw values."
)


# ================================================================
# 10. MODEL EXPLAINABILITY
# ================================================================

print("\n========== MODEL EXPLAINABILITY ==========")

coefficient_table = pd.DataFrame({
    "Feature": active_features,
    "Coefficient": [
        final_model.params[feature]
        for feature in active_features
    ],
    "P-value": [
        final_model.pvalues[feature]
        for feature in active_features
    ]
})

coefficient_table["Significance"] = np.where(
    coefficient_table["P-value"] < SIGNIFICANCE_LEVEL,
    "Significant",
    "Not significant"
)

coefficient_table["Effect Direction"] = np.where(
    coefficient_table["Coefficient"] >= 0,
    "Positive",
    "Negative"
)

coefficient_table = coefficient_table.sort_values(
    by="Coefficient",
    key=lambda s: s.abs(),
    ascending=False
)

display(coefficient_table)

print(
    "\nInterpretation guide:\n"
    "- Positive coefficient: associated with a higher predicted price.\n"
    "- Negative coefficient: associated with a lower predicted price.\n"
    "- Binary coefficient: estimated difference between 1 and 0, "
    "holding other predictors constant.\n"
    "- Scaled numerical coefficient: estimated change across the "
    "training-set minimum-to-maximum range of that feature.\n"
    "- Furnishing dummy coefficients are relative to the omitted "
    "reference category, furnished.\n"
)

# Plot the final coefficients
plt.figure(figsize=(11, 6))

plot_data = coefficient_table.sort_values("Coefficient")

plt.barh(
    plot_data["Feature"],
    plot_data["Coefficient"]
)

plt.axvline(0, linestyle="--")
plt.title("Final Model: Feature Coefficients")
plt.xlabel("Coefficient in Original Price Units")
plt.ylabel("Feature")
plt.tight_layout()
plt.show()


# ================================================================
# 11. HOUSE PRICE CALCULATOR UTILITY FUNCTION
# ================================================================

# This function accepts RAW feature values.
# It performs encoding and scaling automatically.
# It then predicts price using the final OLS model.

def predict_house_price(
    area,
    bedrooms,
    bathrooms,
    stories,
    mainroad,
    guestroom,
    basement,
    hotwaterheating,
    airconditioning,
    parking,
    prefarea,
    furnishingstatus
):
    """
    Predict a house price using the trained regression model.

    Binary inputs: 'yes' or 'no'
    furnishingstatus: 'furnished', 'semi-furnished', or 'unfurnished'
    Numerical inputs: raw, unscaled values
    """

    binary_inputs = {
        "mainroad": mainroad,
        "guestroom": guestroom,
        "basement": basement,
        "hotwaterheating": hotwaterheating,
        "airconditioning": airconditioning,
        "prefarea": prefarea
    }

    row = {
        "area": area,
        "bedrooms": bedrooms,
        "bathrooms": bathrooms,
        "stories": stories,
        "parking": parking
    }

    # Validate numerical inputs.
    for feature, value in row.items():
        if not np.isfinite(float(value)):
            raise ValueError(
                f"{feature} must be a finite number."
            )

        if float(value) < 0:
            raise ValueError(
                f"{feature} cannot be negative."
            )

    # Validate binary categorical inputs.
    for feature, value in binary_inputs.items():
        value = str(value).strip().lower()

        if value not in ["yes", "no"]:
            raise ValueError(
                f"{feature} must be 'yes' or 'no'."
            )

        row[feature] = 1 if value == "yes" else 0

    # Validate furnishing status.
    furnishingstatus = str(
        furnishingstatus
    ).strip().lower()

    allowed_statuses = [
        "furnished",
        "semi-furnished",
        "unfurnished"
    ]

    if furnishingstatus not in allowed_statuses:
        raise ValueError(
            "furnishingstatus must be furnished, "
            "semi-furnished, or unfurnished."
        )

    # Create the two dummy variables.
    row["furnishingstatus_semi-furnished"] = int(
        furnishingstatus == "semi-furnished"
    )

    row["furnishingstatus_unfurnished"] = int(
        furnishingstatus == "unfurnished"
    )

    # Convert the raw input into a one-row DataFrame.
    input_df = pd.DataFrame([row])

    # Match the original feature order.
    input_df = input_df[feature_columns].astype(float)

    # Apply the scaler fitted on training data.
    input_df.loc[:, scale_columns] = scaler.transform(
        input_df[scale_columns]
    )

    # Keep only the predictors retained in the final model.
    input_final = input_df[active_features]

    # Add the intercept and match the OLS model's exact columns.
    input_final = sm.add_constant(
        input_final,
        has_constant="add"
    )

    input_final = input_final.reindex(
        columns=final_model.model.exog_names,
        fill_value=0
    )

    predicted_price = float(
        final_model.predict(input_final).iloc[0]
    )

    return max(0.0, predicted_price)


# ================================================================
# EXAMPLE: PREDICT THE PRICE OF A NEW HOUSE
# ================================================================

estimated_price = predict_house_price(
    area=6000,
    bedrooms=3,
    bathrooms=2,
    stories=2,
    mainroad="yes",
    guestroom="no",
    basement="yes",
    hotwaterheating="no",
    airconditioning="yes",
    parking=1,
    prefarea="yes",
    furnishingstatus="semi-furnished"
)

print("\n========== HOUSE PRICE CALCULATOR ==========")
print(f"Estimated House Price: {estimated_price:,.2f}")


# ================================================================
# PROJECT COMPLETED
# ================================================================

print("\nHousing Price Prediction Project Completed Successfully!")
print("Algorithm: Multiple Linear Regression")
print("Feature Selection: RFE + Backward Elimination")
print("Statistical Model: Statsmodels OLS")
print("Evaluation: R2, MAE, MSE and RMSE")
```
