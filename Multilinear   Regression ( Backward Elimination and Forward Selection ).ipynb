```python

# ================================================================
# HOUSING PRICE PREDICTION
# MULTIPLE LINEAR REGRESSION
# BACKWARD ELIMINATION + FORWARD SELECTION
# Google Colab - Complete Project Code
# ================================================================


# ================================================================
# 1. IMPORT NECESSARY LIBRARIES
# ================================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import statsmodels.api as sm
import warnings

from IPython.display import display
from google.colab import files

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import (
    r2_score,
    mean_absolute_error,
    mean_squared_error
)

from statsmodels.stats.outliers_influence import variance_inflation_factor

warnings.filterwarnings("ignore")
sns.set_theme(style="whitegrid")

RANDOM_STATE = 100
ALPHA = 0.05

print("Libraries imported successfully!")


# ================================================================
# 2. IMPORT THE DATASET
# ================================================================

# Upload Housing.csv if it is not already available in Colab.
# You can also change DATA_PATH to your uploaded file's path.

DATA_PATH = "/content/Housing.csv"

try:
    housing = pd.read_csv(DATA_PATH)
except FileNotFoundError:
    print("Upload your Housing.csv file:")
    uploaded = files.upload()
    if not uploaded:
        raise ValueError("No dataset was uploaded.")
    DATA_PATH = next(iter(uploaded))
    housing = pd.read_csv(DATA_PATH)

housing.columns = housing.columns.str.strip()

print("\nDataset shape:", housing.shape)
print("\nFirst five rows:")
display(housing.head())

print("\nDataset information:")
housing.info()

print("\nStatistical description:")
display(housing.describe(include="all"))

print("\nMissing values:")
display(housing.isnull().sum())

print("\nDuplicate rows:", housing.duplicated().sum())

if "price" not in housing.columns:
    raise ValueError("The dataset must contain a 'price' column.")

if housing.empty:
    raise ValueError("The dataset is empty.")

housing = housing.drop_duplicates().reset_index(drop=True)
housing = housing.dropna(subset=["price"]).copy()

print("\nDataset shape after removing duplicates and rows")
print("with missing target values:", housing.shape)


# ================================================================
# 3. EXPLORATORY DATA ANALYSIS (EDA)
# ================================================================

numeric_columns = housing.select_dtypes(
    include=np.number
).columns.tolist()

categorical_columns = housing.select_dtypes(
    exclude=np.number
).columns.tolist()

# 3.1 Distribution of house prices

plt.figure(figsize=(9, 5))
sns.histplot(housing["price"], kde=True)
plt.title("Distribution of House Prices")
plt.xlabel("Price")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()


# 3.2 Numerical feature distributions

housing[numeric_columns].hist(
    figsize=(14, 10),
    bins=20,
    edgecolor="black"
)

plt.suptitle("Distribution of Numerical Variables")
plt.tight_layout()
plt.show()


# 3.3 Correlation heatmap

plt.figure(figsize=(11, 8))
sns.heatmap(
    housing[numeric_columns].corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    square=True
)

plt.title("Correlation Heatmap of Numerical Variables")
plt.tight_layout()
plt.show()


# 3.4 Numerical features versus price

predictor_numeric = [
    c for c in numeric_columns if c != "price"
]

fig, axes = plt.subplots(
    2, 3, figsize=(16, 9)
)

for ax, col in zip(axes.flat, predictor_numeric):
    sns.scatterplot(
        data=housing,
        x=col,
        y="price",
        ax=ax,
        alpha=0.7
    )
    ax.set_title(f"{col} vs Price")

for ax in axes.flat[len(predictor_numeric):]:
    ax.set_visible(False)

plt.tight_layout()
plt.show()


# 3.5 Categorical features versus price

for col in categorical_columns:
    plt.figure(figsize=(7, 4))
    sns.boxplot(data=housing, x=col, y="price")
    plt.title(f"{col} vs House Price")
    plt.xticks(rotation=15)
    plt.tight_layout()
    plt.show()


# ================================================================
# 4. DATA PREPROCESSING AND FEATURE ENGINEERING
# ================================================================

# Separate target and predictors.

X_raw = housing.drop(columns=["price"]).copy()
y = housing["price"].astype(float).copy()

# Identify numerical predictors.
num_vars = X_raw.select_dtypes(
    include=np.number
).columns.tolist()

# Identify categorical predictors.
cat_vars = X_raw.select_dtypes(
    exclude=np.number
).columns.tolist()

print("\nNumerical predictors:", num_vars)
print("Categorical predictors:", cat_vars)

# Split BEFORE fitting the scaler or selecting features.

X_train_raw, X_test_raw, y_train, y_test = train_test_split(
    X_raw,
    y,
    test_size=0.30,
    random_state=RANDOM_STATE
)

# Encode categorical variables using one-hot encoding.
# drop_first=True prevents the dummy-variable trap.

X_train = pd.get_dummies(
    X_train_raw,
    columns=cat_vars,
    drop_first=True,
    dtype=int
)

X_test = pd.get_dummies(
    X_test_raw,
    columns=cat_vars,
    drop_first=True,
    dtype=int
)

# Ensure training and test columns match exactly.
X_test = X_test.reindex(
    columns=X_train.columns,
    fill_value=0
)

# Convert all features to numeric values.
X_train = X_train.astype(float)
X_test = X_test.astype(float)

# Fit Min-Max scaling ONLY on training numerical predictors.
# This avoids using test-set information during preprocessing.

scaler = MinMaxScaler()

X_train.loc[:, num_vars] = scaler.fit_transform(
    X_train[num_vars]
)

X_test.loc[:, num_vars] = scaler.transform(
    X_test[num_vars]
)

# Preserve the feature names and order.
feature_columns = X_train.columns.tolist()

# Make the data safe for statsmodels.
X_train = X_train.replace([np.inf, -np.inf], np.nan)
X_test = X_test.replace([np.inf, -np.inf], np.nan)

if X_train.isnull().any().any() or X_test.isnull().any().any():
    raise ValueError(
        "Missing or infinite predictor values detected. "
        "Add an imputation step before fitting the model."
    )

print("\nTraining feature shape:", X_train.shape)
print("Testing feature shape:", X_test.shape)

print("\nProcessed training data:")
display(X_train.head())


# ================================================================
# 5. INITIAL MULTIPLE LINEAR REGRESSION MODEL
# ================================================================

X_train_const = sm.add_constant(
    X_train,
    has_constant="add"
)

initial_model = sm.OLS(
    y_train,
    X_train_const
).fit()

print("\nINITIAL OLS MODEL SUMMARY")
print(initial_model.summary())


# ================================================================
# 6. BACKWARD ELIMINATION
# ================================================================

def backward_elimination(X, y, alpha=0.05):
    """
    Remove the predictor with the highest p-value above alpha.
    Repeat until all remaining predictors have p-values <= alpha,
    or no predictors remain.
    """

    selected = list(X.columns)
    history = []

    while selected:
        X_current = sm.add_constant(
            X[selected],
            has_constant="add"
        )

        model = sm.OLS(y, X_current).fit()

        pvalues = model.pvalues.drop(
            labels=["const"],
            errors="ignore"
        )

        if pvalues.empty:
            break

        worst_feature = pvalues.idxmax()
        worst_pvalue = float(pvalues[worst_feature])

        if worst_pvalue <= alpha:
            break

        history.append({
            "Removed Feature": worst_feature,
            "P-value": worst_pvalue,
            "Adjusted R2 Before Removal": model.rsquared_adj
        })

        print(
            f"Removing {worst_feature}: "
            f"p-value = {worst_pvalue:.6f}"
        )

        selected.remove(worst_feature)

    if not selected:
        raise ValueError(
            "Backward elimination removed all predictors."
        )

    final_model = sm.OLS(
        y,
        sm.add_constant(
            X[selected],
            has_constant="add"
        )
    ).fit()

    return selected, final_model, pd.DataFrame(history)


backward_features, backward_model, backward_history = (
    backward_elimination(
        X_train,
        y_train,
        alpha=ALPHA
    )
)

print("\nBACKWARD ELIMINATION FEATURES:")
print(backward_features)

print("\nBACKWARD ELIMINATION SUMMARY:")
print(backward_model.summary())

print("\nBackward elimination history:")
if not backward_history.empty:
    display(backward_history)
else:
    print("No predictors needed to be removed.")


# ================================================================
# 7. FORWARD SELECTION USING ADJUSTED R-SQUARED
# ================================================================

def forward_selection(X, y, min_improvement=1e-8):
    """
    Start with an empty feature set.

    Add the candidate predictor producing the largest adjusted
    R-squared improvement. Stop when no candidate improves
    adjusted R-squared by more than min_improvement.
    """

    remaining = list(X.columns)
    selected = []
    history = []

    current_adj_r2 = -np.inf

    while remaining:
        candidate_results = []

        for candidate in remaining:
            trial_features = selected + [candidate]

            X_trial = sm.add_constant(
                X[trial_features],
                has_constant="add"
            )

            model = sm.OLS(y, X_trial).fit()

            candidate_results.append(
                (model.rsquared_adj, candidate, model)
            )

        candidate_results.sort(
            key=lambda item: item[0],
            reverse=True
        )

        best_adj_r2, best_feature, best_model = (
            candidate_results[0]
        )

        improvement = (
            best_adj_r2 - current_adj_r2
            if np.isfinite(current_adj_r2)
            else np.inf
        )

        if (
            np.isfinite(current_adj_r2)
            and improvement <= min_improvement
        ):
            break

        selected.append(best_feature)
        remaining.remove(best_feature)
        current_adj_r2 = best_adj_r2

        history.append({
            "Added Feature": best_feature,
            "Adjusted R2": best_adj_r2,
            "Improvement": improvement
        })

        print(
            f"Added {best_feature}: "
            f"Adjusted R2 = {best_adj_r2:.6f}"
        )

    if not selected:
        raise ValueError(
            "Forward selection did not select any predictors."
        )

    final_model = sm.OLS(
        y,
        sm.add_constant(
            X[selected],
            has_constant="add"
        )
    ).fit()

    return selected, final_model, pd.DataFrame(history)


forward_features, forward_model, forward_history = (
    forward_selection(
        X_train,
        y_train
    )
)

print("\nFORWARD SELECTION FEATURES:")
print(forward_features)

print("\nFORWARD SELECTION SUMMARY:")
print(forward_model.summary())

print("\nForward selection history:")
display(forward_history)


# ================================================================
# 8. COMPARE THE TWO FEATURE SELECTION MODELS
# ================================================================

selection_comparison = pd.DataFrame({
    "Method": [
        "Backward Elimination",
        "Forward Selection"
    ],
    "Number of Features": [
        len(backward_features),
        len(forward_features)
    ],
    "Training R2": [
        backward_model.rsquared,
        forward_model.rsquared
    ],
    "Training Adjusted R2": [
        backward_model.rsquared_adj,
        forward_model.rsquared_adj
    ],
    "AIC": [
        backward_model.aic,
        forward_model.aic
    ],
    "BIC": [
        backward_model.bic,
        forward_model.bic
    ]
})

print("\nFEATURE SELECTION COMPARISON:")
display(selection_comparison)

# Choose the model with the higher training adjusted R-squared.
# This is the stated selection rule, not a guarantee of the
# lowest test-set prediction error.

if backward_model.rsquared_adj >= forward_model.rsquared_adj:
    selected_method = "Backward Elimination"
    selected_features = backward_features.copy()
    final_model = backward_model
else:
    selected_method = "Forward Selection"
    selected_features = forward_features.copy()
    final_model = forward_model

print("\nSELECTED METHOD:", selected_method)
print("SELECTED FEATURES:", selected_features)


# ================================================================
# 9. VIF ANALYSIS FOR THE SELECTED MODEL
# ================================================================

def calculate_vif(X):
    """
    Calculate VIF for each predictor.
    A constant column is added before calculation.
    """

    X_const = sm.add_constant(
        X,
        has_constant="add"
    )

    vif_rows = []

    for i, col in enumerate(X_const.columns):
        if col == "const":
            continue

        try:
            vif_value = variance_inflation_factor(
                X_const.values,
                i
            )
        except Exception:
            vif_value = np.inf

        vif_rows.append({
            "Feature": col,
            "VIF": vif_value
        })

    return pd.DataFrame(vif_rows).sort_values(
        "VIF",
        ascending=False
    )


vif_table = calculate_vif(
    X_train[selected_features]
)

print("\nVIF TABLE FOR SELECTED FEATURES:")
display(vif_table)

print(
    "\nInterpretation: VIF values above 5 can indicate "
    "potential multicollinearity; values above 10 are "
    "often considered particularly concerning. These are "
    "guidelines, not absolute rules."
)


# ================================================================
# 10. RESIDUAL ANALYSIS ON TRAINING DATA
# ================================================================

X_train_final = sm.add_constant(
    X_train[selected_features],
    has_constant="add"
)

train_predictions = final_model.predict(X_train_final)

train_residuals = y_train - train_predictions

# 10.1 Residuals versus fitted values

plt.figure(figsize=(9, 5))
sns.scatterplot(
    x=train_predictions,
    y=train_residuals,
    alpha=0.7
)

plt.axhline(
    y=0,
    linestyle="--",
    color="red"
)

plt.xlabel("Predicted Price")
plt.ylabel("Residual")
plt.title("Residuals vs Fitted Values")
plt.tight_layout()
plt.show()


# 10.2 Residual distribution

plt.figure(figsize=(9, 5))
sns.histplot(
    train_residuals,
    kde=True
)

plt.xlabel("Residual")
plt.title("Distribution of Training Residuals")
plt.tight_layout()
plt.show()


# 10.3 Normal Q-Q plot

sm.qqplot(
    train_residuals,
    line="45",
    fit=True
)

plt.title("Normal Q-Q Plot of Training Residuals")
plt.tight_layout()
plt.show()


# 10.4 Residuals versus each selected predictor

for col in selected_features:
    plt.figure(figsize=(7, 4))

    sns.scatterplot(
        x=X_train[col],
        y=train_residuals,
        alpha=0.7
    )

    plt.axhline(
        y=0,
        linestyle="--",
        color="red"
    )

    plt.xlabel(col)
    plt.ylabel("Residual")
    plt.title(f"Residuals vs {col}")
    plt.tight_layout()
    plt.show()


# 10.5 Durbin-Watson statistic

from statsmodels.stats.stattools import durbin_watson

dw_stat = durbin_watson(train_residuals)

print("\nDurbin-Watson statistic:", round(dw_stat, 4))

print(
    "A value near 2 suggests little first-order residual "
    "autocorrelation. This statistic alone does not prove "
    "independence."
)


# ================================================================
# 11. PREDICTIONS ON UNSEEN TEST DATA
# ================================================================

X_test_final = sm.add_constant(
    X_test[selected_features],
    has_constant="add"
)

test_predictions = final_model.predict(X_test_final)

# Keep actual and predicted values together.

results = pd.DataFrame({
    "Actual Price": y_test,
    "Predicted Price": test_predictions
})

results["Residual"] = (
    results["Actual Price"] - results["Predicted Price"]
)

print("\nFIRST 15 TEST PREDICTIONS:")
display(results.head(15))


# ================================================================
# 12. MODEL EVALUATION
# ================================================================

def calculate_metrics(y_true, y_pred, number_of_features):
    n = len(y_true)
    p = number_of_features

    r2 = r2_score(y_true, y_pred)

    # Adjusted R-squared for a test set.
    # It is undefined when n <= p + 1.
    if n > p + 1:
        adjusted_r2 = (
            1 - (1 - r2) * (n - 1) / (n - p - 1)
        )
    else:
        adjusted_r2 = np.nan

    rmse = np.sqrt(
        mean_squared_error(y_true, y_pred)
    )

    mae = mean_absolute_error(y_true, y_pred)

    return {
        "R2": r2,
        "Adjusted R2": adjusted_r2,
        "RMSE": rmse,
        "MAE": mae
    }


train_metrics = calculate_metrics(
    y_train,
    train_predictions,
    len(selected_features)
)

test_metrics = calculate_metrics(
    y_test,
    test_predictions,
    len(selected_features)
)

metrics_table = pd.DataFrame(
    [train_metrics, test_metrics],
    index=["Training Data", "Testing Data"]
)

print("\nMODEL PERFORMANCE:")
display(metrics_table)

print("\nFinal model training R-squared:",
      round(final_model.rsquared, 4))

print("Final model training adjusted R-squared:",
      round(final_model.rsquared_adj, 4))

print("Test R-squared:",
      round(test_metrics["R2"], 4))

print("Test RMSE:",
      round(test_metrics["RMSE"], 2))

print("Test MAE:",
      round(test_metrics["MAE"], 2))


# ================================================================
# 13. VISUAL MODEL EVALUATION
# ================================================================

# 13.1 Actual versus predicted prices

plt.figure(figsize=(8, 6))

sns.scatterplot(
    x=y_test,
    y=test_predictions,
    alpha=0.7
)

lower = min(y_test.min(), test_predictions.min())
upper = max(y_test.max(), test_predictions.max())

plt.plot(
    [lower, upper],
    [lower, upper],
    linestyle="--",
    color="red"
)

plt.xlabel("Actual Price")
plt.ylabel("Predicted Price")
plt.title("Actual vs Predicted House Prices")
plt.tight_layout()
plt.show()


# 13.2 Comparison for the first 30 test observations

comparison = results.head(30)

plt.figure(figsize=(13, 6))

plt.plot(
    range(len(comparison)),
    comparison["Actual Price"].values,
    marker="o",
    label="Actual Price"
)

plt.plot(
    range(len(comparison)),
    comparison["Predicted Price"].values,
    marker="x",
    label="Predicted Price"
)

plt.xlabel("Test Observation")
plt.ylabel("Price")
plt.title("Actual vs Predicted Prices: First 30 Test Observations")
plt.legend()
plt.tight_layout()
plt.show()


# 13.3 Test residual plot

plt.figure(figsize=(9, 5))

sns.scatterplot(
    x=test_predictions,
    y=results["Residual"],
    alpha=0.7
)

plt.axhline(
    y=0,
    linestyle="--",
    color="red"
)

plt.xlabel("Predicted Price")
plt.ylabel("Test Residual")
plt.title("Test Residuals vs Predicted Prices")
plt.tight_layout()
plt.show()


# ================================================================
# 14. FINAL REGRESSION EQUATION
# ================================================================

print("\nFINAL MULTIPLE LINEAR REGRESSION EQUATION")
print("=" * 60)

print(f"Price = {final_model.params['const']:,.4f}")

for feature in selected_features:
    coefficient = final_model.params[feature]

    sign = "+" if coefficient >= 0 else "-"

    print(
        f"       {sign} {abs(coefficient):,.4f}"
        f" * {feature}"
    )

print("=" * 60)

print(
    "\nIMPORTANT: The equation above uses Min-Max-scaled "
    "numerical predictors and one-hot-encoded categorical "
    "predictors. The input to the equation must therefore "
    "use the same preprocessing."
)


# ================================================================
# 15. MODEL EXPLAINABILITY
# ================================================================

coefficient_table = pd.DataFrame({
    "Feature": final_model.params.index,
    "Coefficient": final_model.params.values,
    "P-value": final_model.pvalues.values
})

coefficient_table = coefficient_table[
    coefficient_table["Feature"] != "const"
].copy()

coefficient_table["Absolute Coefficient"] = (
    coefficient_table["Coefficient"].abs()
)

coefficient_table = coefficient_table.sort_values(
    "Absolute Coefficient",
    ascending=False
)

print("\nCOEFFICIENT AND SIGNIFICANCE TABLE:")
display(coefficient_table)

# Coefficients are not directly comparable across numerical
# features and binary indicators because numerical features
# were scaled but binary indicators were not.

plt.figure(figsize=(10, 6))

sns.barplot(
    data=coefficient_table,
    x="Coefficient",
    y="Feature"
)

plt.axvline(
    x=0,
    linestyle="--",
    color="black"
)

plt.title("Selected Model Coefficients")
plt.xlabel("Coefficient")
plt.ylabel("Feature")
plt.tight_layout()
plt.show()

print("\nINTERPRETATION GUIDE:")
print("1. Positive coefficient: higher feature value predicts")
print("   a higher price, holding other selected features fixed.")
print("2. Negative coefficient: higher feature value predicts")
print("   a lower price, holding other selected features fixed.")
print("3. A p-value below 0.05 indicates statistical significance")
print("   under the fitted OLS model's assumptions.")
print("4. Coefficients describe associations, not necessarily")
print("   causal effects.")


# ================================================================
# 16. PRICE CALCULATOR UTILITY FUNCTION
# ================================================================

# The calculator accepts the original, unscaled house features.
# It automatically applies the same encoding and scaling used
# for model training.

def predict_house_price(
    area,
    bedrooms,
    bathrooms,
    stories,
    mainroad,
    guestroom,
    basement,
    hotwaterheating,
    airconditioning,
    parking,
    prefarea,
    furnishingstatus
):
    """
    Predict house price from original house attributes.

    Binary categorical inputs: 'yes' or 'no'
    furnishingstatus: 'furnished', 'semi-furnished', 'unfurnished'

    Numerical inputs are in the original dataset's units.
    """

    raw_input = pd.DataFrame([{
        "area": area,
        "bedrooms": bedrooms,
        "bathrooms": bathrooms,
        "stories": stories,
        "mainroad": mainroad,
        "guestroom": guestroom,
        "basement": basement,
        "hotwaterheating": hotwaterheating,
        "airconditioning": airconditioning,
        "parking": parking,
        "prefarea": prefarea,
        "furnishingstatus": furnishingstatus
    }])

    # Validate categorical inputs.
    binary_columns = [
        "mainroad",
        "guestroom",
        "basement",
        "hotwaterheating",
        "airconditioning",
        "prefarea"
    ]

    for col in binary_columns:
        raw_input[col] = (
            raw_input[col]
            .astype(str)
            .str.strip()
            .str.lower()
        )

        if raw_input[col].iloc[0] not in ["yes", "no"]:
            raise ValueError(
                f"{col} must be 'yes' or 'no'."
            )

    raw_input["furnishingstatus"] = (
        raw_input["furnishingstatus"]
        .astype(str)
        .str.strip()
        .str.lower()
    )

    valid_furnishing = [
        "furnished",
        "semi-furnished",
        "unfurnished"
    ]

    if raw_input["furnishingstatus"].iloc[0] not in valid_furnishing:
        raise ValueError(
            "furnishingstatus must be furnished, "
            "semi-furnished, or unfurnished."
        )

    # Validate numerical inputs.
    numerical_input = [
        "area", "bedrooms", "bathrooms",
        "stories", "parking"
    ]

    for col in numerical_input:
        raw_input[col] = pd.to_numeric(
            raw_input[col],
            errors="raise"
        )

        value = raw_input[col].iloc[0]

        if not np.isfinite(value) or value < 0:
            raise ValueError(
                f"{col} must be a finite non-negative number."
            )

    # Encode the single input row.
    input_encoded = pd.get_dummies(
        raw_input,
        columns=cat_vars,
        drop_first=True,
        dtype=int
    )

    # Align exactly with the original training feature columns.
    input_encoded = input_encoded.reindex(
        columns=feature_columns,
        fill_value=0
    ).astype(float)

    # Apply the already-fitted scaler to numerical predictors.
    input_encoded.loc[:, num_vars] = scaler.transform(
        input_encoded[num_vars]
    )

    # Select the final model's predictors in the correct order.
    input_final = sm.add_constant(
        input_encoded[selected_features],
        has_constant="add"
    )

    predicted_price = float(
        final_model.predict(input_final).iloc[0]
    )

    return predicted_price


# ================================================================
# 17. EXAMPLE PRICE PREDICTION
# ================================================================

predicted_price = predict_house_price(
    area=6000,
    bedrooms=3,
    bathrooms=2,
    stories=2,
    mainroad="yes",
    guestroom="no",
    basement="yes",
    hotwaterheating="no",
    airconditioning="yes",
    parking=1,
    prefarea="yes",
    furnishingstatus="semi-furnished"
)

print("\nEXAMPLE HOUSE PRICE PREDICTION")
print("-" * 40)
print(f"Estimated house price: {predicted_price:,.2f}")
print("-" * 40)


# ================================================================
# 18. INTERACTIVE PRICE CALCULATOR
# ================================================================

def interactive_price_calculator():
    """
    Ask the user for house attributes and predict the price.
    """

    print("\nHOUSING PRICE CALCULATOR")
    print("=" * 40)

    try:
        area = float(input("Enter area: "))
        bedrooms = int(input("Enter number of bedrooms: "))
        bathrooms = int(input("Enter number of bathrooms: "))
        stories = int(input("Enter number of stories: "))

        mainroad = input(
            "Main road access (yes/no): "
        ).strip().lower()

        guestroom = input(
            "Guestroom (yes/no): "
        ).strip().lower()

        basement = input(
            "Basement (yes/no): "
        ).strip().lower()

        hotwaterheating = input(
            "Hot water heating (yes/no): "
        ).strip().lower()

        airconditioning = input(
            "Air conditioning (yes/no): "
        ).strip().lower()

        parking = int(input("Enter parking spaces: "))

        prefarea = input(
            "Preferred area (yes/no): "
        ).strip().lower()

        furnishingstatus = input(
            "Furnishing status "
            "(furnished/semi-furnished/unfurnished): "
        ).strip().lower()

        price_prediction = predict_house_price(
            area=area,
            bedrooms=bedrooms,
            bathrooms=bathrooms,
            stories=stories,
            mainroad=mainroad,
            guestroom=guestroom,
            basement=basement,
            hotwaterheating=hotwaterheating,
            airconditioning=airconditioning,
            parking=parking,
            prefarea=prefarea,
            furnishingstatus=furnishingstatus
        )

        print("\nPREDICTED HOUSE PRICE")
        print("=" * 40)
        print(f"{price_prediction:,.2f}")
        print("=" * 40)

    except (ValueError, TypeError) as error:
        print("Invalid input:", error)


# Uncomment the next line to run the interactive calculator
# interactive_price_calculator()


# ================================================================
# 19. SAVE MODEL RESULTS
# ================================================================

results.to_csv(
    "housing_price_test_predictions.csv",
    index=True
)

selection_comparison.to_csv(
    "feature_selection_comparison.csv",
    index=False
)

coefficient_table.to_csv(
    "housing_model_coefficients.csv",
    index=False
)

print("\nSaved files:")
print("1. housing_price_test_predictions.csv")
print("2. feature_selection_comparison.csv")
print("3. housing_model_coefficients.csv")

# Optional: download the prediction results from Colab.
# Uncomment to download the file automatically.

# files.download("housing_price_test_predictions.csv")

print("\nPROJECT EXECUTION COMPLETED SUCCESSFULLY!")
```
