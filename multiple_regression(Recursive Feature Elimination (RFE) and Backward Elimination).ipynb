```python
# ================================================================
# PROJECT: MULTIPLE LINEAR REGRESSION
# HOUSING PRICE PREDICTION
# Google Colab - Complete Corrected Code
# Dataset: Housing.csv
# ================================================================


# STEP 1: IMPORT LIBRARIES
# ================================================================

import os
import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import statsmodels.api as sm

from google.colab import files
from IPython.display import display

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.feature_selection import RFE
from sklearn.metrics import (
    r2_score,
    mean_absolute_error,
    mean_squared_error
)

sns.set_theme(style="whitegrid")

print("All libraries imported successfully!")


# STEP 2: LOAD DATASET
# ================================================================

file_path = "/content/Housing.csv"

if not os.path.exists(file_path):
    print("Please upload your Housing.csv file.")
    uploaded = files.upload()

    csv_files = [
        name for name in uploaded
        if name.lower().endswith(".csv")
    ]

    if not csv_files:
        raise FileNotFoundError(
            "No CSV file uploaded. Please upload Housing.csv."
        )

    file_path = csv_files[0]

df = pd.read_csv(file_path)

# Standardize column names
df.columns = df.columns.str.strip().str.lower()

print("\nDATASET LOADED SUCCESSFULLY")
print("Dataset shape:", df.shape)

print("\nFirst five records:")
display(df.head())

print("\nColumn names:")
print(df.columns.tolist())

print("\nDataset information:")
df.info()

if "price" not in df.columns:
    raise ValueError(
        "The dataset must contain a column named 'price'."
    )


# STEP 3: DATA CLEANING
# ================================================================

# Remove duplicate rows
df = df.drop_duplicates().copy()

# Convert target to numeric
df["price"] = pd.to_numeric(
    df["price"], errors="coerce"
)

# Remove records with missing target values
df = df.dropna(subset=["price"]).copy()

# Convert numeric columns safely
for col in df.columns:
    if col != "price":
        converted = pd.to_numeric(df[col], errors="coerce")

        # Convert only if the column is already numeric
        # or its nonmissing values are all numeric
        original_nonmissing = df[col].notna().sum()

        if (
            pd.api.types.is_numeric_dtype(df[col])
            or (
                original_nonmissing > 0
                and converted.notna().sum() == original_nonmissing
            )
        ):
            df[col] = converted

# Fill missing numerical values with their medians
numeric_cols = df.select_dtypes(
    include=np.number
).columns.tolist()

for col in numeric_cols:
    df[col] = df[col].fillna(df[col].median())

# Fill missing categorical values with their modes
categorical_cols = df.select_dtypes(
    exclude=np.number
).columns.tolist()

for col in categorical_cols:
    mode_values = df[col].mode()

    if not mode_values.empty:
        df[col] = df[col].fillna(mode_values.iloc[0])

# Remove constant input columns
constant_cols = [
    col for col in df.columns
    if col != "price" and df[col].nunique() <= 1
]

df.drop(columns=constant_cols, inplace=True)

print("\nDATA CLEANING COMPLETED")
print("Cleaned dataset shape:", df.shape)

print("\nMissing values after cleaning:")
print(df.isnull().sum())

print("\nDuplicate rows remaining:", df.duplicated().sum())


# STEP 4: EXPLORATORY DATA ANALYSIS (EDA)
# ================================================================

print("\nSTATISTICAL SUMMARY")
display(df.describe(include="all"))

# 4.1 Distribution of house prices
plt.figure(figsize=(8, 5))
sns.histplot(df["price"], kde=True)
plt.title("Distribution of Housing Prices")
plt.xlabel("Price")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()

# 4.2 House price versus area
if "area" in df.columns:
    plt.figure(figsize=(8, 5))
    sns.scatterplot(data=df, x="area", y="price")
    plt.title("House Price vs Area")
    plt.xlabel("Area")
    plt.ylabel("Price")
    plt.tight_layout()
    plt.show()

# 4.3 Average price by furnishing status
if "furnishingstatus" in df.columns:
    plt.figure(figsize=(8, 5))
    sns.barplot(
        data=df,
        x="furnishingstatus",
        y="price",
        errorbar=None
    )
    plt.title("Average Price by Furnishing Status")
    plt.xlabel("Furnishing Status")
    plt.ylabel("Average Price")
    plt.tight_layout()
    plt.show()

# 4.4 Numerical correlation heatmap
plt.figure(figsize=(10, 7))
sns.heatmap(
    df.select_dtypes(include=np.number).corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()


# STEP 5: SEPARATE INPUT FEATURES AND TARGET
# ================================================================

X_raw = df.drop(columns=["price"]).copy()
y = df["price"].astype(float).copy()

# Remember original categorical columns and their categories
categorical_features = X_raw.select_dtypes(
    exclude=np.number
).columns.tolist()

category_levels = {
    col: sorted(X_raw[col].dropna().unique().tolist())
    for col in categorical_features
}

# Convert categorical variables into dummy variables
X = pd.get_dummies(
    X_raw,
    columns=categorical_features,
    drop_first=True,
    dtype=int
)

# Ensure numerical predictors are numeric
X = X.astype(float)

print("\nFEATURES AFTER DUMMY ENCODING")
print(X.columns.tolist())
print("Number of predictors:", X.shape[1])

if X.shape[1] == 0:
    raise ValueError("No usable predictor columns were found.")


# STEP 6: SPLIT DATA INTO TRAINING AND TESTING SETS
# ================================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("\nTRAIN-TEST SPLIT")
print("Training records:", len(X_train))
print("Testing records:", len(X_test))


# STEP 7: RECURSIVE FEATURE ELIMINATION (RFE)
# ================================================================

n_features = min(8, X_train.shape[1])

rfe = RFE(
    estimator=LinearRegression(),
    n_features_to_select=n_features,
    step=1
)

rfe.fit(X_train, y_train)

rfe_features = X_train.columns[
    rfe.support_
].tolist()

print("\nFEATURES SELECTED BY RFE")

for feature in rfe_features:
    print(feature)

rfe_ranking = pd.DataFrame({
    "Feature": X_train.columns,
    "Ranking": rfe.ranking_,
    "Selected": rfe.support_
}).sort_values("Ranking")

display(rfe_ranking)


# STEP 8: BACKWARD ELIMINATION USING OLS
# ================================================================

# Backward elimination removes the predictor with the
# highest p-value above 0.05 until all remaining predictors
# have p-values <= 0.05, or only one predictor remains.

selected_features = rfe_features.copy()
alpha = 0.05

while len(selected_features) > 1:

    X_ols = sm.add_constant(
        X_train[selected_features],
        has_constant="add"
    )

    ols_model = sm.OLS(
        y_train,
        X_ols
    ).fit()

    p_values = ols_model.pvalues.drop(
        labels=["const"],
        errors="ignore"
    )

    if p_values.empty:
        break

    worst_feature = p_values.idxmax()
    worst_p_value = p_values[worst_feature]

    if worst_p_value > alpha:
        print(
            f"Removing {worst_feature} "
            f"(p-value = {worst_p_value:.4f})"
        )
        selected_features.remove(worst_feature)
    else:
        break

print("\nFINAL SELECTED FEATURES")

for feature in selected_features:
    print(feature)

# Final OLS model for statistical summary
X_train_ols = sm.add_constant(
    X_train[selected_features],
    has_constant="add"
)

final_ols = sm.OLS(
    y_train,
    X_train_ols
).fit()

print("\nFINAL OLS MODEL SUMMARY")
print(final_ols.summary())


# STEP 9: TRAIN FINAL LINEAR REGRESSION MODEL
# ================================================================

model = LinearRegression()

model.fit(
    X_train[selected_features],
    y_train
)

print("\nFinal regression model trained successfully!")


# STEP 10: RESIDUAL ANALYSIS
# ================================================================

train_predictions = model.predict(
    X_train[selected_features]
)

train_residuals = (
    y_train.to_numpy() - train_predictions
)

# 10.1 Actual versus predicted prices
plt.figure(figsize=(7, 5))
plt.scatter(
    y_train,
    train_predictions,
    alpha=0.7
)

low = min(y_train.min(), train_predictions.min())
high = max(y_train.max(), train_predictions.max())

plt.plot(
    [low, high],
    [low, high],
    "r--",
    label="Perfect prediction"
)

plt.xlabel("Actual Price")
plt.ylabel("Predicted Price")
plt.title("Training Data: Actual vs Predicted")
plt.legend()
plt.tight_layout()
plt.show()

# 10.2 Residuals versus fitted values
plt.figure(figsize=(7, 5))
plt.scatter(
    train_predictions,
    train_residuals,
    alpha=0.7
)

plt.axhline(0, color="red", linestyle="--")
plt.xlabel("Predicted Price")
plt.ylabel("Residual")
plt.title("Residuals vs Fitted Values")
plt.tight_layout()
plt.show()

# 10.3 Residual distribution
plt.figure(figsize=(7, 5))
sns.histplot(train_residuals, kde=True)
plt.xlabel("Residual")
plt.title("Distribution of Residuals")
plt.tight_layout()
plt.show()

# 10.4 Normal Q-Q plot
sm.qqplot(
    train_residuals,
    line="45",
    fit=True
)

plt.title("Normal Q-Q Plot of Residuals")
plt.tight_layout()
plt.show()

# 10.5 Residual statistics
print("\nRESIDUAL DIAGNOSTICS")
print(
    "Mean residual:",
    round(np.mean(train_residuals), 2)
)

print(
    "Residual standard deviation:",
    round(np.std(train_residuals), 2)
)

print(
    "Durbin-Watson statistic:",
    round(
        sm.stats.stattools.durbin_watson(
            train_residuals
        ),
        3
    )
)


# STEP 11: TEST DATA MODEL EVALUATION
# ================================================================

test_predictions = model.predict(
    X_test[selected_features]
)

r2 = r2_score(y_test, test_predictions)

mae = mean_absolute_error(
    y_test,
    test_predictions
)

mse = mean_squared_error(
    y_test,
    test_predictions
)

rmse = np.sqrt(mse)

# MAPE
actual_values = y_test.to_numpy()
nonzero_mask = actual_values != 0

if np.any(nonzero_mask):
    mape = np.mean(
        np.abs(
            (
                actual_values[nonzero_mask]
                - test_predictions[nonzero_mask]
            )
            / actual_values[nonzero_mask]
        )
    ) * 100
else:
    mape = np.nan

print("\n========== MODEL EVALUATION ==========")
print(f"R-squared (R²): {r2:.4f}")
print(f"Mean Absolute Error (MAE): {mae:,.2f}")
print(f"Mean Squared Error (MSE): {mse:,.2f}")
print(f"Root Mean Squared Error (RMSE): {rmse:,.2f}")
print(f"Mean Absolute Percentage Error (MAPE): {mape:.2f}%")

# Compare actual and predicted values
comparison = pd.DataFrame({
    "Actual Price": actual_values,
    "Predicted Price": test_predictions,
    "Absolute Error": np.abs(
        actual_values - test_predictions
    )
})

print("\nFIRST 10 TEST PREDICTIONS")
display(comparison.head(10))

# Test data evaluation plot
plt.figure(figsize=(7, 5))
plt.scatter(
    actual_values,
    test_predictions,
    alpha=0.7
)

low = min(actual_values.min(), test_predictions.min())
high = max(actual_values.max(), test_predictions.max())

plt.plot(
    [low, high],
    [low, high],
    "r--",
    label="Perfect prediction"
)

plt.xlabel("Actual Price")
plt.ylabel("Predicted Price")
plt.title("Test Data: Actual vs Predicted Prices")
plt.legend()
plt.tight_layout()
plt.show()


# STEP 12: DISPLAY FINAL REGRESSION EQUATION
# ================================================================

intercept = model.intercept_
coefficients = model.coef_

print("\n========== FINAL REGRESSION EQUATION ==========")

equation = f"Price = {intercept:,.2f}"

for feature, coefficient in zip(
    selected_features,
    coefficients
):
    sign = "+" if coefficient >= 0 else "-"
    equation += (
        f" {sign} {abs(coefficient):,.2f} * {feature}"
    )

print(equation)


# STEP 13: FEATURE IMPORTANCE
# ================================================================

importance = pd.DataFrame({
    "Feature": selected_features,
    "Coefficient": coefficients,
    "Absolute Impact": np.abs(coefficients)
})

importance = importance.sort_values(
    by="Absolute Impact",
    ascending=False
)

print("\n========== FEATURE IMPORTANCE ==========")
display(importance)

plt.figure(figsize=(9, 5))
sns.barplot(
    data=importance,
    x="Coefficient",
    y="Feature"
)

plt.title("Regression Coefficients")
plt.xlabel("Coefficient")
plt.ylabel("Feature")
plt.tight_layout()
plt.show()

print(
    "\nInterpretation: Each coefficient estimates the change "
    "in predicted price for a one-unit increase in that feature, "
    "holding other selected features constant."
)


# STEP 14: PREPARE A NEW HOUSE FOR PREDICTION
# ================================================================

new_house = pd.DataFrame([{
    "area": 5000,
    "bedrooms": 3,
    "bathrooms": 2,
    "stories": 2,
    "mainroad": "yes",
    "guestroom": "no",
    "basement": "yes",
    "hotwaterheating": "no",
    "airconditioning": "yes",
    "parking": 1,
    "prefarea": "yes",
    "furnishingstatus": "semi-furnished"
}])

# Make sure input columns match the dataset
new_house = new_house.reindex(
    columns=X_raw.columns
)

# Apply the training dataset's category levels
for col in categorical_features:
    if col in new_house.columns:
        new_house[col] = pd.Categorical(
            new_house[col],
            categories=category_levels[col]
        )

# Encode using the same convention as the training data
new_house_encoded = pd.get_dummies(
    new_house,
    columns=categorical_features,
    drop_first=True,
    dtype=int
)

# Align columns with training features
new_house_encoded = new_house_encoded.reindex(
    columns=X.columns,
    fill_value=0
).astype(float)

# Predict using the final selected features
new_house_final = new_house_encoded[
    selected_features
]

predicted_price = model.predict(
    new_house_final
)[0]

print("\n========== HOUSE PRICE PREDICTION ==========")
print("Input house details:")
display(new_house)

print(
    f"Estimated house price: {predicted_price:,.2f}"
)


# STEP 15: REUSABLE HOUSE PRICE CALCULATOR
# ================================================================

def predict_house_price(
    area,
    bedrooms,
    bathrooms,
    stories,
    mainroad,
    guestroom,
    basement,
    hotwaterheating,
    airconditioning,
    parking,
    prefarea,
    furnishingstatus
):
    """
    Predict the price of a house using the trained model.

    Categorical values must match the dataset categories.
    """

    house = pd.DataFrame([{
        "area": area,
        "bedrooms": bedrooms,
        "bathrooms": bathrooms,
        "stories": stories,
        "mainroad": mainroad,
        "guestroom": guestroom,
        "basement": basement,
        "hotwaterheating": hotwaterheating,
        "airconditioning": airconditioning,
        "parking": parking,
        "prefarea": prefarea,
        "furnishingstatus": furnishingstatus
    }])

    # Keep the same input columns as the training dataset
    house = house.reindex(
        columns=X_raw.columns
    )

    # Check and apply categorical levels
    for col in categorical_features:
        if col in house.columns:
            value = house[col].iloc[0]

            if value not in category_levels[col]:
                raise ValueError(
                    f"Invalid value '{value}' for {col}. "
                    f"Allowed values: {category_levels[col]}"
                )

            house[col] = pd.Categorical(
                house[col],
                categories=category_levels[col]
            )

    # Dummy encode
    encoded_house = pd.get_dummies(
        house,
        columns=categorical_features,
        drop_first=True,
        dtype=int
    )

    # Match training columns exactly
    encoded_house = encoded_house.reindex(
        columns=X.columns,
        fill_value=0
    ).astype(float)

    # Predict
    prediction = model.predict(
        encoded_house[selected_features]
    )[0]

    return prediction


# EXAMPLE: CHANGE VALUES TO PREDICT ANOTHER HOUSE
# ================================================================

example_price = predict_house_price(
    area=6000,
    bedrooms=4,
    bathrooms=2,
    stories=2,
    mainroad="yes",
    guestroom="no",
    basement="yes",
    hotwaterheating="no",
    airconditioning="yes",
    parking=2,
    prefarea="yes",
    furnishingstatus="furnished"
)

print("\n========== PRICE CALCULATOR RESULT ==========")
print(f"Estimated price: {example_price:,.2f}")


# STEP 16: FINAL PROJECT SUMMARY
# ================================================================

print("\n" + "=" * 55)
print("HOUSING PRICE PREDICTION PROJECT COMPLETED")
print("=" * 55)

print("Dataset records:", len(df))
print("Training records:", len(X_train))
print("Testing records:", len(X_test))
print("Initial encoded features:", X.shape[1])
print("Final selected features:", len(selected_features))
print("Selected feature names:", selected_features)
print(f"Test R-squared: {r2:.4f}")
print(f"Test MAE: {mae:,.2f}")
print(f"Test RMSE: {rmse:,.2f}")

print("\nAll project steps executed successfully.")
```
